# Cell, nucleus, and nuclear-IOD data

Path24 contains finalized reviewed image objects. Relative nuclear IOD is an image-derived phenotype, not an independently validated absolute genome size. Genomic and microscopy records are joined only at species level.

This notebook reads only the compact canonical tables in this analysis directory. Change or add cells freely; the paper figures are built separately in R/ggplot.

```python
from pathlib import Path
import pandas as pd

ROOT = next(
    candidate for candidate in (Path.cwd(), *Path.cwd().parents)
    if (candidate / 'pyproject.toml').exists()
)
pd.set_option('display.max_columns', 100)
```

```python
DATA = ROOT / 'analyses/02_morphology/data'
panel = pd.read_csv(ROOT / 'data/identity/path24_panel.csv')
objects = pd.read_csv(DATA / 'cell_nucleus_objects.csv', low_memory=False)
species = pd.read_csv(DATA / 'cell_nucleus_species_estimates.csv')
iod_objects = pd.read_csv(DATA / 'nuclear_iod_objects.csv', low_memory=False)
iod_species = pd.read_csv(DATA / 'relative_nuclear_iod_species.csv')

assert (panel['species'].nunique(), objects.shape[0], iod_objects.shape[0]) == (24, 1152, 805)
{'Path24 species': panel['species'].nunique(), 'reviewed cell-nucleus objects': len(objects), 'reviewed IOD nuclei': len(iod_objects)}
```

```python
species.merge(iod_species, on='species', suffixes=('_morphology', '_iod'))
```
