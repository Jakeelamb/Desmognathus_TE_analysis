# Phylogenetic path data

The final fitted model is the three-trait Path24 analysis: relative nuclear IOD, nucleus area, and cell area. It is exploratory and does not uniquely identify causal arrow direction. SVL and life-history tables are retained for inspection but are not predictors in this fitted Path24 model.

This notebook reads only the compact canonical tables in this analysis directory. Change or add cells freely; the paper figures are built separately in R/ggplot.

```python
from pathlib import Path
import pandas as pd

ROOT = next(
    candidate for candidate in (Path.cwd(), *Path.cwd().parents)
    if (candidate / 'pyproject.toml').exists()
)
pd.set_option('display.max_columns', 100)
```

```python
DATA = ROOT / 'analyses/03_phylogenetic_path/data'
traits = pd.read_csv(DATA / 'path24_traits.csv')
organismal = pd.read_csv(DATA / 'organismal_traits.csv')
models = pd.read_csv(DATA / 'path_model_comparison.csv')
edges = pd.read_csv(DATA / 'path_edge_estimates.csv')
overlap = pd.read_csv(DATA / 'te_relative_iod_overlap.csv')

assert (traits['tree_tip'].nunique(), len(overlap)) == (24, 21)
{'Path24 species': traits['tree_tip'].nunique(), 'TE-IOD overlap': len(overlap), 'candidate classes': len(models)}
```

```python
organismal_for_join = organismal.rename(columns={'species': 'organismal_species'})
traits.merge(organismal_for_join, left_on='tree_tip', right_on='organismal_species', how='left', validate='one_to_one')[[
'species', 'relative_iod_index', 'nucleus_area_um2', 'cell_area_um2', 'max_svl_mm', 'reproductive_strategy']].sort_values('species')
```
